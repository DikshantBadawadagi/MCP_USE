```python
import asyncio
from langchain_google_genai import ChatGoogleGenerativeAI
from mcp_use import MCPAgent, MCPClient

config = {
    "mcpServers" : {
        "brave-search" : {
            "command" : "npx",
            "args" : ["-y","@smithery/cli@latest", "run", "@smithery-ai/brave-search","--config", "{\"braveApiKey\" : \"1234567\"}"]
        }
    }
}

client  = MCPClient.from_dict(config)

llm = ChatGoogleGenerativeAI(model="gemini-2.5-pro-exp-03-25",google_api_key="")

agent = MCPAgent(llm=llm, client=client, max_steps=30)

async def run_query():
  result = await agent.run("Give me the latest news on US Tariffs on China")
  print(f"\nResult: {result}")

await run_query()

```
